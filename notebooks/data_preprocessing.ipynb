```python
# 1. Import libraries
import pandas as pd
import numpy as np

# ===============================
# 2. Load dataset
# ===============================
df = pd.read_csv("../data/heart_disease.csv", header=None)

# According to UCI docs, assign column names
columns = [
    "age", "sex", "cp", "trestbps", "chol",
    "fbs", "restecg", "thalach", "exang",
    "oldpeak", "slope", "ca", "thal", "target"
]
df.columns = columns

print("Dataset shape:", df.shape)
df.head()

# ===============================
# 3. Replace "?" with NaN and clean
# ===============================
df = df.replace("?", np.nan)

# Convert numeric columns properly
for col in df.columns:
    df[col] = pd.to_numeric(df[col], errors="ignore")

print("Missing values per column:")
print(df.isnull().sum())

# Fill missing values (simple strategy: mode for categorical, median for numerical)
for col in df.columns:
    if df[col].dtype == "object":
        df[col] = df[col].fillna(df[col].mode()[0])
    else:
        df[col] = df[col].fillna(df[col].median())

# ===============================
# 4. Convert target into binary
# ===============================
df["target"] = df["target"].apply(lambda x: 1 if x > 0 else 0)

print("\nTarget value counts:")
print(df["target"].value_counts())

# ===============================
# 5. Save cleaned dataset
# ===============================
df.to_csv("../data/heart_disease_cleaned.csv", index=False)
print("\n✅ Cleaned dataset saved to ../data/heart_disease_cleaned.csv")
```
